```python
import requests
import pandas as pd

url = "https://dogapi.dog/api/v2/breeds"

all_breeds = []

# To get all the breeds we have to get the 10 pages of data.

for page in range(1, 11):

    response = requests.get(
        url,
        params={"page[number]": page},
        timeout=30
    )

    response.raise_for_status()
    data = response.json()

    all_breeds.extend(data["data"])

    print(f"Page {page} downloaded")

print("Total breeds:", len(all_breeds))
```

```python
breeds = []

for dog in data["data"]:
    attrs = dog["attributes"]

    breeds.append({
        "breed_id": dog["id"],
        "breed": attrs.get("name"),
        "life_min": attrs.get("life", {}).get("min"),
        "life_max": attrs.get("life", {}).get("max"),
        "male_weight_min": attrs.get(
            "male_weight", {}
        ).get("min"),
        "male_weight_max": attrs.get(
            "male_weight", {}
        ).get("max"),
        "female_weight_min": attrs.get(
            "female_weight", {}
        ).get("min"),
        "female_weight_max": attrs.get(
            "female_weight", {}
        ).get("max"),
        "hypoallergenic": attrs.get("hypoallergenic")
    })

df_dogs = pd.DataFrame(breeds)

display(df_dogs.head())
```

```python
df_dogs = pd.json_normalize(all_breeds)

df_dogs = df_dogs.drop_duplicates(subset=["id"])

print("Total unique breeds:", len(df_dogs))

display(df_dogs.head())
```

```python
df_dogs.to_csv("dog_breeds_complete.csv", index=False, encoding="utf-8-sig")

# To download our .csv from our API
```
